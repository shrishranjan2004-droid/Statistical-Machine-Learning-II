```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the Rosenbrock Objective Function (Minimization)
def rosenbrock(x):
    # x is expected to be a 2D array of shape (n_particles, 2) or a 1D array of shape (2,)
    if x.ndim == 1:
        return (1 - x[0])**2 + 100.0 * (x[1] - x[0]**2)**2
    else:
        return (1 - x[:, 0])**2 + 100.0 * (x[:, 1] - x[:, 0]**2)**2

def pso_rosenbrock(n_particles=30, dimensions=2, iterations=150, w=0.5, c1=1.5, c2=1.5):
    # Initialize particle positions randomly within bounds [-2, 2]
    np.random.seed(42)
    x = np.random.uniform(-2.0, 2.0, (n_particles, dimensions))
    
    v = np.zeros((n_particles, dimensions))
    
   
    p_best = x.copy()
    p_best_fitness = rosenbrock(x)
    
    g_best_idx = np.argmin(p_best_fitness)
    g_best = p_best[g_best_idx].copy()
    g_best_fitness = p_best_fitness[g_best_idx]
    
    trajectory = [x.copy()]
    
    print(f"Initial Best Fitness: {g_best_fitness:.6f} at Position {g_best}")
    
    # Main PSO Loop
    for t in range(1, iterations + 1):
        r1 = np.random.rand(n_particles, dimensions)
        r2 = np.random.rand(n_particles, dimensions)
        
        # Velocity Update Equation with Inertia (w), Cognitive (c1), and Social (c2) components
        v = w * v + c1 * r1 * (p_best - x) + c2 * r2 * (g_best - x)
        
        # Position Update Equation
        x = x + v
        
        # current fitness
        fitness = rosenbrock(x)
        
        improved = fitness < p_best_fitness
        p_best[improved] = x[improved]
        p_best_fitness[improved] = fitness[improved]
        
        
        current_g_best_idx = np.argmin(p_best_fitness)
        if p_best_fitness[current_g_best_idx] < g_best_fitness:
            g_best = p_best[current_g_best_idx].copy()
            g_best_fitness = p_best_fitness[current_g_best_idx]
            
        trajectory.append(x.copy())
        
        if t % 25 == 0 or t == 1:
            print(f"Iteration {t:3d} | Best Fitness: {g_best_fitness:.6e} | Best Position: {g_best}")
            
    print(out := f"\nOptimization Complete!\nGlobal Minimum Found at: {g_best}\nFinal Fitness Value: {g_best_fitness:.6e}")
    return np.array(trajectory), g_best


trajectory, best_position = pso_rosenbrock()
```
