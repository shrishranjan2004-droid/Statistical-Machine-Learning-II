```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.decomposition import KernelPCA, PCA

X, y = make_circles(n_samples=500, factor=0.3, noise=0.05, random_state=42)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

kpca = KernelPCA(n_components=2, kernel="rbf", gamma=15)
X_kpca = kpca.fit_transform(X)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].scatter(X[y == 0, 0], X[y == 0, 1], color='red', alpha=0.7, label='Class 0 (Inner)')
axes[0].scatter(X[y == 1, 0], X[y == 1, 1], color='blue', alpha=0.7, label='Class 1 (Outer)')
axes[0].set_title("Original Data (Non-Linear)")
axes[0].set_xlabel("Feature 1")
axes[0].set_ylabel("Feature 2")
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

axes[1].scatter(X_pca[y == 0, 0], X_pca[y == 0, 1], color='red', alpha=0.7, label='Class 0')
axes[1].scatter(X_pca[y == 1, 0], X_pca[y == 1, 1], color='blue', alpha=0.7, label='Class 1')
axes[1].set_title("Standard PCA (Fails to Separate)")
axes[1].set_xlabel("Principal Component 1")
axes[1].set_ylabel("Principal Component 2")
axes[1].grid(True, linestyle='--', alpha=0.5)

axes[2].scatter(X_kpca[y == 0, 0], X_kpca[y == 0, 1], color='red', alpha=0.7, label='Class 0')
axes[2].scatter(X_kpca[y == 1, 0], X_kpca[y == 1, 1], color='blue', alpha=0.7, label='Class 1')
axes[2].set_title("Kernel PCA with RBF Kernel (Separated)")
axes[2].set_xlabel("Kernel Principal Component 1")
axes[2].set_ylabel("Kernel Principal Component 2")
axes[2].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
